```python
from google.colab import userdata
tavily_key=userdata.get("TAVILY")
gemini_key=userdata.get("gemini_key")
```

```python
from tavily import TavilyClient

def web_search(query):
  tavily_client=TavilyClient(tavily_key)
  response=tavily_client.search(query)

  content=""
  for result in response["results"]:
    content+=result["content"]+"\n\n"
  return content
```

```python
topic="Latest AI agents in 2026"
```

```python
from google import genai
client=genai.Client(api_key=gemini_key)

def summarizer_agent(results):
  response=client.models.generate_content(
      model="gemini-2.5-flash",
      contents=f"Summarize the following in 5 bullet points:{results}"
  )
  return response.text
```

```python
results=web_search(topic)
summary=summarizer_agent(results)
print(summary)
```

```python
def facebook_agent(summary):

    prompt = f"""
    Act as a professional social media content writer.

    I will provide you with a summary. Based on that summary, create an engaging Facebook post.

    Instructions:
    - Start with a strong hook in the first 1-2 lines.
    - Use simple and clear language.
    - Make it conversational and relatable.
    - Add emotional connection where possible.
    - Keep it between 150-300 words.
    - Break into small paragraphs for readability.
    - Add relevant emojis (but do not overuse them).
    - End with a question or call-to-action to increase engagement.
    - Do not repeat the summary word-for-word. Rewrite creatively.

    Here is the summary:

    {summary}
    """

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text

```

```python
facebook_post=facebook_agent(summary)
print(facebook_post)
```
